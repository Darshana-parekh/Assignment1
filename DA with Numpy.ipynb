```python
#ASSESSMENT 2
import numpy as np
import pandas as pd
df=pd.read_csv('C:/Users/Admin/Downloads/historical_automobile_sales.csv',encoding='latin-1')
```

```python
df
```

```python
#Q1.
import matplotlib.pyplot as plt

# Group data by Year and calculate total automobile sales
yearly_sales = df.groupby('Year')['Automobile_Sales'].sum()

# Create a line chart
plt.figure(figsize=(10, 5))
yearly_sales.plot(kind='line', marker='o')

# Customize the chart
plt.title("Automobile Sales by Year")
plt.xlabel("Year")
plt.ylabel("Total Automobile Sales")
plt.grid(True)

# Display the chart
plt.show()
```

```python
#q2

# Filter data for recession periods
recession_data = df[df['Recession'] == 1]

# Group by Year and Vehicle Type
sales_by_vehicle = recession_data.groupby(['Year', 'Vehicle_Type'])['Automobile_Sales'].sum().unstack()

# Plot line chart
plt.figure(figsize=(12, 6))
sales_by_vehicle.plot(marker='o')

plt.title("Automobile Sales Trends by Vehicle Type During Recession Periods")
plt.xlabel("Year")
plt.ylabel("Automobile Sales")
plt.grid(True)
plt.legend(title="Vehicle Type")
plt.tight_layout()

plt.show()
```

```python
#Q3.
import seaborn as sns


# Group data by Recession and Vehicle Type
sales_trend = (
    df.groupby(['Recession', 'Vehicle_Type'])['Automobile_Sales']
      .mean()
      .reset_index()
)

# Convert Recession values to labels
sales_trend['Recession'] = sales_trend['Recession'].map({
    0: 'Non-Recession',
    1: 'Recession'
})

# Create the visualization
plt.figure(figsize=(12,6))

sns.barplot(
    data=sales_trend,
    x='Vehicle_Type',
    y='Automobile_Sales',
    hue='Recession',
    palette='Set2'
)

plt.title("Average Automobile Sales by Vehicle Type\n(Recession vs Non-Recession)")
plt.xlabel("Vehicle Type")
plt.ylabel("Average Automobile Sales")
plt.xticks(rotation=45)
plt.legend(title="Period")
plt.tight_layout()

plt.show()
```

```python
# Q 4

# Average sales by Vehicle Type and Recession status
vehicle_sales = (
    df.groupby(['Vehicle_Type', 'Recession'])['Automobile_Sales']
      .mean()
      .reset_index()
)

# Convert Recession values to labels
vehicle_sales['Recession'] = vehicle_sales['Recession'].map({
    0: 'Non-Recession',
    1: 'Recession'
})

# Plot grouped bar chart
plt.figure(figsize=(12,6))

sns.barplot(
    data=vehicle_sales,
    x='Vehicle_Type',
    y='Automobile_Sales',
    hue='Recession',
    palette='viridis'
)

plt.title("Comparison of Vehicle Sales During Recession and Non-Recession Periods")
plt.xlabel("Vehicle Type")
plt.ylabel("Average Automobile Sales")
plt.xticks(rotation=45)
plt.legend(title="Period")
plt.tight_layout()

plt.show()
```
